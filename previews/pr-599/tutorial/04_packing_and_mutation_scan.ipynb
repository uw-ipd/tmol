We want to change a small part of a protein while controlling what the rest can do. Using ten ubiquitin residues, we will finish with a mutation comparison and three sequence-design outcomes, each with explicit packing restrictions.

**We will:**

1. Repack residues 3–8 while keeping the sequence fixed.
2. Compare mutations at residue 5 with a repacked wild-type control.
3. Allow residues 5 and 7 to choose their identities together, then check the result.

Afterward, an optional section shows the candidate side-chain conformations behind the search. Start with [scoring](03_scoring_and_analysis.ipynb).

```python
import logging

# The example files do not need local PDB or CCD mirrors.
logging.getLogger("atomworks.constants").addFilter(
    lambda record: not record.getMessage().startswith(
        (
            "Environment variable CCD_MIRROR_PATH not set.",
            "Environment variable PDB_MIRROR_PATH not set.",
        )
    )
)

from contextlib import redirect_stderr, redirect_stdout
from io import StringIO
from pathlib import Path
from time import perf_counter
import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import torch
from IPython.display import display
from biotite.structure.io import load_structure

import tmol
from tmol.database import ParameterDatabase
from tmol.io import pose_stack_from_biotite
from tmol.numeric import coord_dihedrals
from tmol.ops import (
    build_sidechain_coord_mask,
    compute_block_adjacency,
    compute_block_centroids_and_furthest_dist,
    res_mask_to_coord_mask,
)
from tmol.pack import PackerPalette, PackerTask, SetPackerTask, pack_rotamers
from tmol.pack.rotamer import (
    FixedAAChiSampler,
    IncludeCurrentSampler,
    build_rotamers,
)
from tmol.pack.rotamer.dunbrack import create_dunbrack_sampler_from_database
from tmol.pose import PoseStackBuilder
from tmol.score import beta2016_score_function

SEED = 20260807
np.random.seed(SEED)
torch.manual_seed(SEED)
warnings.filterwarnings(
    "ignore", message=r"Sparse invariant checks are implicitly disabled.*"
)

device = (
    torch.device("cuda", torch.cuda.current_device())
    if tmol.has_cuda() and torch.cuda.is_available()
    else torch.device("cpu")
)
repo_root = next(
    (
        path
        for path in (Path.cwd(), *Path.cwd().parents)
        if (path / "tmol/tests/data").is_dir()
    ),
    Path(tmol.__file__).resolve().parents[1],
)
if not (repo_root / "tmol/tests/data/cif/1UBQ.cif").exists():
    repo_root = Path(tmol.__file__).resolve().parents[1]
cif_path = repo_root / "tmol/tests/data/cif/1UBQ.cif"

from itables import show as show_table

# Format displayed numbers without changing the values used in calculations.
pd.options.display.float_format = "{:.4g}".format


def total_score(pose):
    scorer = score_function.render_whole_pose_scoring_module(pose)
    return float(scorer(pose.coords).detach().cpu()[0])
```

## Load the ten-residue example

Select residues 1–10 of chain A, build the pose, and create the score function. Keep the pose and score function on the same device. Hydrogen optimization is enabled before we compare scores.


```python
param_db = ParameterDatabase.get_default()
atom_array = load_structure(str(cif_path), model=1, include_bonds=True)
protein_slice = atom_array[
    (atom_array.chain_id == "A") & (atom_array.res_id <= 10)
]
with redirect_stdout(StringIO()), redirect_stderr(StringIO()):
    # Optimize hydrogen coordinates before comparing interpreted scores.
    pose_stack = pose_stack_from_biotite(
        protein_slice, device, param_db=param_db, no_optH=False
    )
score_function = beta2016_score_function(device, param_db=param_db)


print(
    f"device={device}; input={cif_path.name}; slice blocks={pose_stack.max_n_blocks}"
)
```

## Repack residues 3–8

Repacking chooses side-chain conformations, called **rotamers**, while keeping the amino-acid sequence and backbone fixed. Here only residues 3–8 of a ten-residue ubiquitin fragment may repack.

A `PackerPalette` supplies compatible residue types; a `PackerTask` narrows the allowed changes. Restrict it to repacking, select zero-based blocks `2:8`, and disable packing everywhere else. The three samplers supply library rotamers, the single ideal conformer for alanine or glycine, and the input conformation.


```python
def add_standard_samplers(task):
    task.add_conformer_sampler(
        create_dunbrack_sampler_from_database(param_db, device)
    )
    task.add_conformer_sampler(FixedAAChiSampler())
    task.add_conformer_sampler(IncludeCurrentSampler())
```

### Define the task

The mask is `True` where side chains may change. Invert it for `disable_packing_by_block_mask()`.


```python
palette = PackerPalette()
repack_task = PackerTask(pose_stack, palette)
repack_task.restrict_to_repacking()

packing_region = torch.zeros_like(pose_stack.block_type_ind, dtype=torch.bool)
packing_region[:, 2:8] = True
repack_task.disable_packing_by_block_mask(~packing_region)
add_standard_samplers(repack_task)
```

### Pack and rescore

`pack_rotamers()` returns a new pose. `total_score()` renders a fresh scorer for it, since packing can change atom layout. Lower values are more favorable under this score function.


```python
start_score = total_score(pose_stack)
torch.manual_seed(SEED)
repacked_pose = pack_rotamers(pose_stack, score_function, repack_task)
repacked_score = total_score(repacked_pose)
```

### Measure the movement

Measure only side-chain atoms in the selected region; report RMS displacement, maximum movement, and the number moving more than 0.001 Å.


```python
packing_atom_mask = res_mask_to_coord_mask(pose_stack, packing_region)
packing_sidechain_mask = packing_atom_mask & build_sidechain_coord_mask(
    pose_stack
)
sidechain_displacement = torch.linalg.vector_norm(
    repacked_pose.coords - pose_stack.coords, dim=-1
)[packing_sidechain_mask]
sidechain_rms_displacement = float(
    torch.sqrt(torch.mean(sidechain_displacement.square())).detach().cpu()
)
max_sidechain_displacement = float(sidechain_displacement.max().detach().cpu())
changed_sidechain_atoms = int(
    (sidechain_displacement > 1e-3).sum().detach().cpu()
)

repack_frame = pd.DataFrame(
    [
        {
            "stage": "input",
            "beta2016": start_score,
            "score_change": 0.0,
            "sidechain_RMSD_A": 0.0,
            "max_movement_A": 0.0,
            "changed_sidechain_atoms": 0,
        },
        {
            "stage": "regional repack",
            "beta2016": repacked_score,
            "score_change": repacked_score - start_score,
            "sidechain_RMSD_A": sidechain_rms_displacement,
            "max_movement_A": max_sidechain_displacement,
            "changed_sidechain_atoms": changed_sidechain_atoms,
        },
    ]
)
show_table(
    repack_frame[
        [
            "stage",
            "beta2016",
            "score_change",
            "sidechain_RMSD_A",
            "max_movement_A",
            "changed_sidechain_atoms",
        ]
    ].rename(
        columns={
            "stage": "Stage",
            "beta2016": "Score",
            "score_change": "Score Δ",
            "sidechain_RMSD_A": "RMS motion (Å)",
            "max_movement_A": "Max motion (Å)",
            "changed_sidechain_atoms": "Moved atoms",
        }
    )
)
```

Compare **input** and **regional repack** in the viewer. Look for moving side chains on a stationary backbone. The table reports their score change and displacement.

Packing is a stochastic search. CPU and CUDA use different search algorithms, so the same seed need not produce the same result across devices.


```python
display(
    tmol.switchable_view(
        {"input": pose_stack, "regional repack": repacked_pose},
        notes={
            "input": f"weighted score: {start_score:.3f}",
            "regional repack": (
                f"score {repacked_score:.3f}; Δ {repacked_score - start_score:+.3f}; "
                f"side-chain RMS motion {sidechain_rms_displacement:.3f} Å"
            ),
        },
        width=720,
        height=420,
    )
)
```

## Compare mutations at one residue

Return to the original ten-residue pose for each mutation. Replace residue 5 (block 4) with ALA, LEU, or PHE. Include its original identity as a **repacked wild-type control**. Let nearby residues repack in every case so each variant has the same opportunity to adjust.

The helper below restricts the target to one requested identity per batch member and keeps all other identities unchanged. `restrict_absent_name3s()` keeps the requested names at the selected blocks. Like `restrict_to_repacking()`, it only removes choices.


```python
def restrict_batched_substitutions(task, block_index, name3_by_pose):
    """Request one target identity per pose and keep other identities fixed."""
    target = torch.zeros_like(task.is_real_block)
    target[:, block_index] = True
    task.restrict_to_repacking(~target)
    if len(name3_by_pose) != target.shape[0]:
        raise ValueError("Provide exactly one requested identity per pose")
    for pose_index, name3 in enumerate(name3_by_pose):
        selected = torch.zeros_like(target)
        selected[pose_index, block_index] = True
        task.restrict_absent_name3s({name3}, selected)
```

```python
target_block = 4
original_type_index = int(pose_stack.block_type_ind64[0, target_block].item())
original_name3 = pose_stack.packed_block_types.active_block_types[
    original_type_index
].name3
variants = list(dict.fromkeys([original_name3, "ALA", "LEU", "PHE"]))
target_label = (
    f"{pose_stack.pdb_info.chain_labels[0, target_block]}:"
    f"{pose_stack.pdb_info.residue_labels[0, target_block]}"
    f"{pose_stack.pdb_info.residue_insertion_codes[0, target_block]}"
)
```

### Choose the local repacking shell

Use residue enclosing spheres plus a 5 Å gap to select neighbors of the target. The printed author labels identify the shell.


```python
# Build one explicit local shell with the existing score_utils geometry helpers.
block_centroids, block_radii = compute_block_centroids_and_furthest_dist(
    pose_stack
)
block_adjacency = compute_block_adjacency(block_centroids, block_radii)
mutation_shell = block_adjacency[:, target_block].clone()
mutation_shell[:, target_block] = True
shell_block_indices = (
    torch.nonzero(mutation_shell[0], as_tuple=False).flatten().tolist()
)
shell_labels = [
    f"{pose_stack.pdb_info.chain_labels[0, block]}:"
    f"{pose_stack.pdb_info.residue_labels[0, block]}"
    f"{pose_stack.pdb_info.residue_insertion_codes[0, block]}"
    for block in shell_block_indices
]
print(
    f"mutation target block {target_block} ({target_label}); local repacking shell "
    f"has {len(shell_block_indices)} blocks: {', '.join(shell_labels)}"
)
```

### Measure target interactions

Sum both matrix orientations between the target and every other residue. Render for the new pose each time because mutations change atom layout.


```python
def target_to_rest_score(pose, block_index):
    scorer = score_function.render_block_pair_scoring_module(pose)
    matrix = scorer(pose.coords, sum_terms=True, apply_weights=True)[0]
    partners = torch.ones(
        matrix.shape[0], dtype=torch.bool, device=matrix.device
    )
    partners[block_index] = False
    return float(
        (
            matrix[block_index, partners].sum()
            + matrix[partners, block_index].sum()
        )
        .detach()
        .cpu()
    )
```

### Batch and pack the variants

Each member requests one identity at the target. Apply the same local shell and samplers to all members.


```python
# CUDA uses two seeded outcomes per identity. On CPU, show one batch member
# per identity to keep execution short.
n_scan_outcomes = 2 if device.type == "cuda" else 1
scan_requests = [
    (outcome, variant)
    for outcome in range(1, n_scan_outcomes + 1)
    for variant in variants
]
requested_variants = [variant for _, variant in scan_requests]
mutation_batch = PoseStackBuilder.from_poses(
    [pose_stack] * len(scan_requests), device
)
mutation_scan_task = PackerTask(mutation_batch, PackerPalette())
restrict_batched_substitutions(
    mutation_scan_task, target_block, requested_variants
)
mutation_shell_batch = mutation_shell.expand(len(scan_requests), -1).clone()
mutation_scan_task.disable_packing_by_block_mask(~mutation_shell_batch)
add_standard_samplers(mutation_scan_task)

torch.manual_seed(
    SEED
)  # Seeds either backend; CPU and CUDA outcomes can differ.
if device.type == "cuda":
    torch.cuda.synchronize()
scan_start = perf_counter()
packed_mutation_batch = pack_rotamers(
    mutation_batch, score_function, mutation_scan_task
)
if device.type == "cuda":
    torch.cuda.synchronize()
scan_seconds = perf_counter() - scan_start
```

### Collect scores and check identities

Check that non-target identities are unchanged, then record each outcome’s total and target-to-rest score.


```python
scan_rows = []
variant_poses = {}
for pose_index, (outcome, variant) in enumerate(scan_requests):
    packed_variant = packed_mutation_batch.split(pose_index)
    state_label = (
        f"{variant} (repacked WT)" if variant == original_name3 else variant
    )
    if outcome == 1:
        variant_poses[state_label] = packed_variant

    assert packed_variant.block_type(0, target_block).name3 == variant

    # The monotonic identity helper must leave every non-target identity unchanged.
    for block_index in range(packed_variant.max_n_blocks):
        if block_index != target_block:
            assert (
                packed_variant.block_type(0, block_index).name3
                == pose_stack.block_type(0, block_index).name3
            )

    scan_rows.append(
        {
            "block_index": target_block,
            "pdb_label": target_label,
            "variant": variant,
            "state": state_label,
            "outcome": outcome,
            "total_score": total_score(packed_variant),
            "interaction_score": target_to_rest_score(
                packed_variant, target_block
            ),
            "shared_batch_seconds": scan_seconds,
        }
    )

scan_frame = pd.DataFrame(scan_rows)
show_table(
    scan_frame.sort_values(["variant", "outcome"])[
        ["variant", "outcome", "total_score", "interaction_score"]
    ].rename(
        columns={
            "variant": "Identity",
            "outcome": "Outcome",
            "total_score": "Total score",
            "interaction_score": "Target–rest score",
        }
    )
)
```

### Compare with repacked wild type

Average each identity’s outcomes, then subtract the WT mean. The searches are independent, so individual outcomes are not paired.


```python
# Summarize each identity independently, then compare distribution means.
scan_summary = scan_frame.groupby(["variant", "state"], as_index=False).agg(
    total_mean=("total_score", "mean"),
    total_min=("total_score", "min"),
    total_max=("total_score", "max"),
    interaction_mean=("interaction_score", "mean"),
    interaction_min=("interaction_score", "min"),
    interaction_max=("interaction_score", "max"),
)
scan_summary["total_range"] = (
    scan_summary["total_max"] - scan_summary["total_min"]
)
scan_summary["interaction_range"] = (
    scan_summary["interaction_max"] - scan_summary["interaction_min"]
)
wt_summary = scan_summary.loc[scan_summary["variant"] == original_name3].iloc[
    0
]
scan_summary["total_delta"] = (
    scan_summary["total_mean"] - wt_summary["total_mean"]
)
scan_summary["interaction_delta"] = (
    scan_summary["interaction_mean"] - wt_summary["interaction_mean"]
)
scan_summary = scan_summary.sort_values("total_delta")
show_table(
    scan_summary[
        [
            "variant",
            "total_mean",
            "total_delta",
            "interaction_mean",
            "interaction_delta",
        ]
    ]
)
```

```python
plot_summary = scan_summary.sort_values("interaction_delta").reset_index(
    drop=True
)
fig, ax = plt.subplots(figsize=(8, 4.2))
for y_position, row in plot_summary.iterrows():
    outcome_offsets = (
        scan_frame.loc[
            scan_frame["variant"] == row["variant"],
            "interaction_score",
        ].to_numpy()
        - wt_summary["interaction_mean"]
    )
    ax.hlines(
        y_position,
        outcome_offsets.min(),
        outcome_offsets.max(),
        color="0.45",
        linewidth=2,
    )
    ax.scatter(
        outcome_offsets,
        np.full_like(outcome_offsets, y_position),
        color="0.35",
    )
    ax.scatter(
        row["interaction_delta"],
        y_position,
        color="tab:blue",
        marker="D",
        zorder=3,
    )
ax.axvline(0.0, color="black", linewidth=1)
ax.set_yticks(np.arange(len(plot_summary)), plot_summary["state"])
ax.set(
    xlabel="Target-to-rest score change from repacked WT (score units)",
    title=f"Local mutation scan at {target_label}: {n_scan_outcomes} outcome(s) per identity",
)
ax.grid(axis="x", alpha=0.25)
plt.tight_layout()
plt.show()
```

The viewer highlights the common repacking shell in orange.


```python
shell_atom_mask = res_mask_to_coord_mask(pose_stack, mutation_shell)
shell_viewer = tmol.view(
    pose_stack,
    highlighted=shell_atom_mask[pose_stack.real_atoms],
    highlight_color="#ff7f0e",
)
print(
    f"Orange: local repacking shell ({len(shell_block_indices)} blocks); "
    f"mutation target: block {target_block} ({target_label})."
)
shell_viewer.show()
```

In the plot, zero is the repacked wild-type mean. Negative values indicate a more favorable target-to-rest interaction score under this protocol. Dots are individual outcomes; diamonds are means.

The orange region is allowed to repack. A single CPU outcome or two CUDA outcomes cannot establish a stable mutation ranking. These score differences are not ΔΔG estimates.


```python
lowest_mean_state = scan_summary.iloc[0]
print(
    "lowest mean total score in this local scan:",
    lowest_mean_state["state"],
)
summary_by_state = scan_summary.set_index("state")
display(
    tmol.switchable_view(
        variant_poses,
        notes={
            state: (
                f"outcome 1 structure; independently summarized mean score "
                f"{summary_by_state.loc[state, 'total_mean']:.3f}; "
                f"observed range "
                f"{summary_by_state.loc[state, 'total_range']:.3f}"
            )
            for state in variant_poses
        },
        width=720,
        height=420,
    )
)
```

## Design two residues together

Allow blocks 4 and 6 (residues 5 and 7) to choose from their original identity, ALA, LEU, or PHE. Their neighbors may repack; everything outside that shell stays fixed.

The role table makes these three rules visible. We run three batch members and compare the sequences and scores the packer selects. They may select the same result.


```python
def restrict_regional_design(task, allowed_name3_by_block):
    """Allow the requested alphabet at each design position."""
    design = torch.zeros_like(task.is_real_block)
    design[:, list(allowed_name3_by_block)] = True
    task.restrict_to_repacking(~design)
    for block_index, names in allowed_name3_by_block.items():
        selected = torch.zeros_like(design)
        selected[:, block_index] = True
        task.restrict_absent_name3s(names, selected)
```

### Choose positions and identities

Use blocks 4 and 6 and keep each original identity as one allowed choice.


```python
design_blocks = [4, 6]
design_alphabet = {
    block_index: {
        pose_stack.block_type(0, block_index).name3,
        "ALA",
        "LEU",
        "PHE",
    }
    for block_index in design_blocks
}

# The union of adjacency neighborhoods defines which blocks may pack.
design_block_mask = torch.zeros_like(
    pose_stack.block_type_ind, dtype=torch.bool
)
design_block_mask[:, design_blocks] = True
design_shell = torch.any(block_adjacency[:, design_blocks, :], dim=1)
design_shell |= design_block_mask
design_repack_only = design_shell & ~design_block_mask
design_frozen = ~design_shell
```

### Show the three roles

Map each block back to its residue label and mark it as designable, repack-only, or frozen.


```python
def design_pdb_label(block_index):
    return (
        f"{pose_stack.pdb_info.chain_labels[0, block_index]}:"
        f"{pose_stack.pdb_info.residue_labels[0, block_index]}"
        f"{pose_stack.pdb_info.residue_insertion_codes[0, block_index]}"
    )


design_role_rows = []
for block_index in range(pose_stack.max_n_blocks):
    if bool(design_block_mask[0, block_index]):
        role = "design"
    elif bool(design_repack_only[0, block_index]):
        role = "repack-only"
    else:
        role = "frozen"
    design_role_rows.append(
        {
            "block_index": block_index,
            "pdb_label": design_pdb_label(block_index),
            "starting_identity": pose_stack.block_type(0, block_index).name3,
            "role": role,
        }
    )
design_role_frame = pd.DataFrame(design_role_rows)
show_table(design_role_frame)
```

### Run three design searches

Each batch member uses the same allowed alphabet and packing shell.


```python
n_design_batch_members = 3
design_batch = PoseStackBuilder.from_poses(
    [pose_stack] * n_design_batch_members, device
)
design_task = PackerTask(design_batch, PackerPalette())
restrict_regional_design(design_task, design_alphabet)
design_shell_batch = design_shell.expand(n_design_batch_members, -1).clone()
design_task.disable_packing_by_block_mask(~design_shell_batch)
add_standard_samplers(design_task)

torch.manual_seed(SEED)
if device.type == "cuda":
    torch.cuda.synchronize()
design_start = perf_counter()
designed_batch = pack_rotamers(design_batch, score_function, design_task)
if device.type == "cuda":
    torch.cuda.synchronize()
design_seconds = perf_counter() - design_start
```

### Check what changed

Mutations can shift atom offsets. Compare each frozen residue’s own coordinates, then report the designed identities and total score.


```python
design_rows = []
design_poses = {}
frozen_block_indices = (
    torch.nonzero(design_frozen[0], as_tuple=False).flatten().tolist()
)
for pose_index in range(n_design_batch_members):
    designed_pose = designed_batch.split(pose_index)
    label = f"batch member {pose_index + 1}"
    identities = [
        designed_pose.block_type(0, block_index).name3
        for block_index in design_blocks
    ]

    for block, identity in zip(design_blocks, identities):
        assert identity in design_alphabet[block]

    # Compare each frozen block in local coordinates; mutations can shift later offsets.
    max_frozen_coordinate_change = 0.0
    for block_index in frozen_block_indices:
        input_block_type = pose_stack.block_type(0, block_index)
        output_block_type = designed_pose.block_type(0, block_index)
        assert output_block_type.name == input_block_type.name
        n_block_atoms = len(input_block_type.atoms)
        input_offset = int(pose_stack.block_coord_offset64[0, block_index])
        output_offset = int(designed_pose.block_coord_offset64[0, block_index])
        input_block_coords = pose_stack.coords[
            0, input_offset : input_offset + n_block_atoms
        ]
        output_block_coords = designed_pose.coords[
            0, output_offset : output_offset + n_block_atoms
        ]
        # Pose reconstruction may introduce a few micro-angstroms of float32
        # roundoff even when packing is disabled for the block.
        torch.testing.assert_close(
            output_block_coords, input_block_coords, rtol=1e-6, atol=1e-5
        )
        block_change = float(
            torch.linalg.vector_norm(
                output_block_coords - input_block_coords, dim=-1
            )
            .max()
            .detach()
            .cpu()
        )
        max_frozen_coordinate_change = max(
            max_frozen_coordinate_change, block_change
        )

    design_poses[label] = designed_pose
    design_rows.append(
        {
            "batch_member": pose_index + 1,
            "block_4_identity": identities[0],
            "block_6_identity": identities[1],
            "sequence_pair": "-".join(identities),
            "total_score": total_score(designed_pose),
            "max_frozen_coordinate_change_A": max_frozen_coordinate_change,
            "shared_batch_seconds": design_seconds,
        }
    )

design_frame = pd.DataFrame(design_rows).sort_values("total_score")
show_table(
    design_frame[
        [
            "batch_member",
            "sequence_pair",
            "total_score",
            "max_frozen_coordinate_change_A",
        ]
    ].rename(
        columns={
            "batch_member": "Search",
            "sequence_pair": "Residues 5 / 7",
            "total_score": "Score",
            "max_frozen_coordinate_change_A": "Max fixed-atom motion (Å)",
        }
    )
)
```

Inspect the allowed region and switch among the three design results.

### Measure candidate χ angles

This helper finds each named torsion’s four atoms and measures its dihedral. It reads rotamer coordinates without running a packing search.


```python
def named_chi_frame(label, rotamer_pose, rotamer_set, block_index):
    rows = []
    n_rotamers = int(rotamer_set.n_rots_for_block[0, block_index])
    rotamer_offset = int(rotamer_set.rot_offset_for_block[0, block_index])
    for local_rotamer in range(n_rotamers):
        rotamer_index = rotamer_offset + local_rotamer
        block_type_index = int(
            rotamer_set.block_type_ind_for_rot[rotamer_index]
        )
        block_type = rotamer_pose.packed_block_types.active_block_types[
            block_type_index
        ]
        coordinate_offset = int(
            rotamer_set.coord_offset_for_rot[rotamer_index]
        )
        row = {"sampling": label, "rotamer": local_rotamer}
        for torsion_index in block_type.sc_torsions:
            atom_indices = block_type.ordered_torsions[torsion_index, :, 0]
            if np.any(atom_indices < 0):
                continue
            coordinates = rotamer_set.coords[
                coordinate_offset
                + torch.as_tensor(
                    atom_indices, dtype=torch.long, device=device
                )
            ].double()
            radians = coord_dihedrals(
                coordinates[0:1],
                coordinates[1:2],
                coordinates[2:3],
                coordinates[3:4],
            )[0]
            torsion_name = block_type.torsions[torsion_index].name
            row[f"{torsion_name}_degrees"] = float(
                torch.rad2deg(radians).cpu()
            )
        rows.append(row)
    return pd.DataFrame(rows)
```

Apply it to Lys6, which has both χ1 and χ2. Each row is one candidate.


```python
# Block 5 is LYS in this 1UBQ slice and therefore demonstrates both chi1 and chi2.
rotamer_inspection_block = 5
rotamer_geometry_frame = pd.concat(
    [
        named_chi_frame(
            "standard",
            standard_rotamer_pose,
            standard_rotamers,
            rotamer_inspection_block,
        ),
        named_chi_frame(
            "expanded",
            expanded_rotamer_pose,
            expanded_rotamers,
            rotamer_inspection_block,
        ),
    ],
    ignore_index=True,
)
show_table(rotamer_geometry_frame)
```

Plot χ1 against χ2, wrapped to −180° to 180°, so neighboring angular samples are easy to compare.


```python
chi1_column = next(
    (
        column
        for column in rotamer_geometry_frame.columns
        if column.lower().startswith("chi1") and column.endswith("_degrees")
    ),
    None,
)
chi2_column = next(
    (
        column
        for column in rotamer_geometry_frame.columns
        if column.lower().startswith("chi2") and column.endswith("_degrees")
    ),
    None,
)
if chi1_column is not None and chi2_column is not None:
    wrapped_geometry = rotamer_geometry_frame.copy()
    wrapped_geometry["chi1_wrapped_degrees"] = (
        wrapped_geometry[chi1_column] + 180.0
    ) % 360.0 - 180.0
    wrapped_geometry["chi2_wrapped_degrees"] = (
        wrapped_geometry[chi2_column] + 180.0
    ) % 360.0 - 180.0
    fig, ax = plt.subplots(figsize=(5.5, 5.0))
    for label, values in wrapped_geometry.groupby("sampling"):
        ax.scatter(
            values["chi1_wrapped_degrees"],
            values["chi2_wrapped_degrees"],
            label=label,
            alpha=0.7,
        )
    ax.set(
        xlim=(-180, 180),
        ylim=(-180, 180),
        xlabel="wrapped χ1 (degrees)",
        ylabel="wrapped χ2 (degrees)",
        title=f"Candidate side-chain geometry at block {rotamer_inspection_block}",
    )
    ax.set_aspect("equal", adjustable="box")
    ax.grid(alpha=0.3)
    ax.legend()
    plt.tight_layout()
    plt.show()
```
